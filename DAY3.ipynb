```python
from datetime import date
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    @classmethod
    def calculate_age(cls,name,birthyear):
            return cls(name,date.today().year-birthyear)

    def show(self):
            print(self.name+"'s age is'"+str(self.age))
age=student('abc',20)
age.show()
```

```python
by=student.calculate_age('abc',2006)
by.show()
```

```python
class coll:
    #class variable
    name="St.Joseph"
    @classmethod
    def school_name(cls):
        print('School name is', cls.name)
    #create class method
    coll.school_name=classmethod(coll.school_name)
#call class method
coll.school_name()
```

```python
#static method
class emp:
    @staticmethod
    def sample(x):
        print('Inside static method',x)
        #call static method
emp.sample(10)
#can be called using object
emp1=emp()
emp1.sample(10)
```

```python
#mini proj
class BankAccount:
    def __init__(self,name,balance=0):
        #instance attributes
        self.name=name
        self.balance=balance
    #deposit money
    def deposit(self,amount):
        if amount>0:
            self.balance+=amount
            print("Deposit Successfully!")
            print("Deposited:",amount)
        else:
            print("Invalid deposit amount")
    #withdaw money
    def withdraw(self,amount):
        if amount <=0:
            print('Invalid withdrawl amount!')
        elif amount > self.balance:
            print('Insufficient balance!')
        else:
            self.balance -=amount
            print("Withdrawal successfull")
            print("Withdrawn:",amount)
    #check balance
    def check_balance(self):
        print("Current balance:",self.balance)
    #display account details
    def display_details(self):
        print("Account Holder:",self.name)
        print("Balance:",self.balance)
        
```

```python
acc1=BankAccount('abc',70000)
acc1.display_details()
```

```python
acc1.deposit(5000)
```

```python
acc1.check_balance()
```

```python
acc1.withdraw(3500)
```

```python
acc1.check_balance()
```

```python
acc1.display_details()
```

```python
#Parent class
class Person:
    def person_info(self,name,age):
        print('Inside Person Class')
        print('Name:',name,'Age:',age)
#parent class 2
class Company:
    def company_info(self,company_name,location):
        print('Inside the Company Class')
        print('Name:',company_name,'Location:',location)
```

```python
#child class
class Employee(Person,Company):
    def Employee_info(self,salary,skill):
        print('Inside employee class - child class')
        print('Salary:',salary,'Skills:',skill)
```

```python
# Create an object of Employee
emp2=Employee()
#access the data
emp2.person_info('abc',20)
emp2.company_info('Edunet Foundation','Bengaluru')
emp2.Employee_info(55000,'AIML Engineer')
```

```python
#Super() - function
class Company:
    def company_name(self):
        return 'Edunet Foundation'
class Employee(Company):
    def info(self):
        #call the super func
        c_name=super().company_name()
        print('ABC works at',c_name)
```

```python
#method overriding
class Vehicle:
    def max_speed(self):
        print("max speed is 100kmph")
class Car(Vehicle):
    #overriding the implementation of vehicle class
    def max_speed(self):
        print('max speed is 200kmph')

car=Car()
car.max_speed()
```

```python
#public member
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    #public instance method
    def show(self):
        #access public data members
        print('Name',self.name,'Salary',self.salary)
```

```python
emp=Employee('ABC',10000)
#access the public data members
print('Name',emp.name,'Salary',emp.salary)
emp.show()
```

```python
#Private member
class Employee:
    def __init__(self,name,salary):
        self.name=name #public data member
        self._salary=salary #private data member
```

```python
#create an object 
emp=Employee('abc',6000)
#access the private data member
print('Salary',emp.__salary)
```

```python
#protected member
class Company:
    def __init__(self):
        #protected member
        self._project="NLP"
#Child class
class Employee(Company):
    def __init__(self,name):
        self.name=name
        Company.__init__(self)
    def show(self):
        print("Employee Name:",self.name)
        #Access the protected member
        print('Working on project:',self._project)
```

```python
c=Employee('ABC')
c.show()
```

```python
#direct access
print('Project:',c._project)
```

```python
class India():
    def capital(self):
        print("New Delhi is the capital of India.")
    def language(self):
        print("India has 22 official languages.")
    def type(self):
        print("India is a developing nation.")
class USA():
    def capital(self):
        print("Capital of USA is Washington DC.")
    def language(self): 
        print("The official language in US is English.")
    def type(self):
        print("USA is a developed nation.")
```

```python
class Car: 
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("Drive!")
class Boat:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("Sail!")
class Plane:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("Fly!")
```

```python
ob_Car=Car('Roma Ferrari','Ferrari')
ob_Boat=Boat('Montauk','Boston Whaler')
ob_Plane=Plane('Boeing 737 ','Boeing')
ob_Car.move()
ob_Boat.move()
ob_Plane.move()
```

```python
#polymorphism with inheritance
class Bird:
    def intro(self):
        print('There are many types of birds!')
    def flight(self):
        print('Most of the birds can fly but some cannot!')
class sparrow(Bird):
    def flight(self):
        print("Sparrows can fly")
class ostrich(Bird):
    def flight(self):
        print("Ostrich cannot fly")
```

```python
ob_bird=Bird()
ob_bird.intro()
ob_bird.flight()
```

```python
ob_sparrow=sparrow()
ob_sparrow.flight()
```

```python
ob_ostrich=ostrich()
ob_ostrich.flight()
```

```python
#overloading
class complex:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def __add__(self,other):
        return self.a+other.a,self.b+other.b
ob1=complex(1,2)
ob2=complex(2,3)
ob3=ob1+ob2
print(ob3)
```
