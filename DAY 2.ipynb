```python
# _init_constructor - used to initialise
class person:
    def __init__(self,name,age,school):
        self.name=name #instance variable
        self.age=age
        self.school=school
    def myfun(self):
        print('Hello, My name is ',self.name)
        print('My age is', self.age)
        print('I went to ',self.school)

p1=person('abc',38,'Geetha School')
p1.myfun()
```

```python
class Person:
    def __init__(god,name,age):
        god.name=name
        god.age=age
    def myfunc(god):
        print('Hello my name is ',god.name)
        print('I AM',god.age)
q1=Person('abc','Immortal')
q1.myfunc()
```

```python
#accessing attributes and methods
class emp:
    name='abc'
    salary='10,00,000'
    def show(self):
        print(self.name)
        print(self.salary)
```

```python
e1=emp()
```

```python
#using getattr instead e1.name
print(getattr(e1,'name'))
```

```python
#hasattr - retun true if the object has attribute
print(hasattr(e1,'name'))
```

```python
print(hasattr(e1,'gender'))
```

```python
#setattr - sets am attribute
setattr(e1,'gender','F')
print(getattr(e1,'gender'))
```

```python
# delete attribute
#delattr - del the attribute
delattr(e1,'gender')
print(getattr(e1,'gender'))
```

```python
class Student:
    def __init__(self,usn,name,age): #constructor method 
        #Instance variable
        self.usn=usn
        self.name=name
        self.age=age
    #instance method access instance variable
    def show(self):
        print('USN:',self.usn,',Name:',self.name,',Age:',self.age)
    #instance method to modify instance variable
    def update(self,usn,age):
        self.usn=usn
        self.age=age
s1=Student('4VV24CS116','Hrishi',20)
s1.show()
```

```python
s1.update('4VV24CS1116',21)
s1.show()
```
